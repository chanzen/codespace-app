```python
from sklearn.cluster import KMeans
import numpy as np

# Example dataset: customers with [Age, Income]
X = np.array([
    [25, 25000],
    [30, 30000],
    [45, 45000],
    [50, 50000],
    [23, 22000],
    [40, 40000],
    [55, 52000],
    [60, 60000]
])

# Create KMeans model with 3 clusters
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(X)

# Cluster assignments for each customer
labels = kmeans.labels_
print("Cluster labels:", labels)

# Centroids of the clusters
centroids = kmeans.cluster_centers_
print("Cluster centroids:\n", centroids)

# Predict cluster for a new customer (Age=35, Income=35000)
new_customer = np.array([[35, 35000]])
cluster = kmeans.predict(new_customer)
print("New customer belongs to cluster:", cluster[0])
```
