```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


dataUrl = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
data = pd.read_csv(dataUrl)
print(data.info())
print(data.describe())

# handle missing values
data['Age'] = data['Age'].fillna(data['Age'].median())
data['Embarked'] = data['Embarked'].fillna(data['Embarked'].mode()[0])

#Remove duplicate rows
data = data.drop_duplicates()

#Filter first class passengers
first_class_passengers = data[data['Pclass'] == 1]
print(first_class_passengers.head())

# Group by
survivedByClass = data.groupby('Pclass')["Survived"].mean()
survivedByClass.plot(kind='bar', title='Survival Rate by Passenger Class', color='skyblue')
plt.xlabel('Passenger Class')
plt.ylabel('Survival Rate')
plt.show()

# Visualization histogram of ages
sns.histplot(data['Age'], bins=30, kde=True)
plt.title('Age Distribution of Titanic Passengers')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()

#Scatter plot of Age vs Fare
plt.scatter(data['Age'], data['Fare'], alpha=0.5, color='orange')
plt.title('Age vs Fare')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.show()
```
