```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
import numpy as np

# Example data: ad clicks vs purchase (1=yes, 0=no)
X = np.array([[1],[2],[3],[6],[7],[8]])
y = np.array([0,0,0,1,1,1])

model = LogisticRegression()
model.fit(X, y)

# Probability of buying after 6 clicks
prob = model.predict_proba([[6]])[0][1]
print("Probability of purchase:", prob)


# Random Forest
rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(X, y)
prob_rf = rf.predict_proba([[5]])[0][1]

# Gradient Boosting
gb = GradientBoostingClassifier(n_estimators=100, random_state=42)
gb.fit(X, y)
prob_gb = gb.predict_proba([[5]])[0][1]

print("Probability of purchase with 5 clicks:")
print("Random Forest:", prob_rf)
print("Gradient Boosting:", prob_gb)
```
