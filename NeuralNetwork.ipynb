```python
from sklearn.neural_network import MLPClassifier
import numpy as np

# Example data: study hours, sleep hours → pass/fail
X = np.array([[2,5],[3,6],[5,4],[6,7]])
y = np.array([0,0,1,1])  # 1=pass, 0=fail

model = MLPClassifier(hidden_layer_sizes=(5,), max_iter=1000, random_state=42)
model.fit(X, y)

# Predict for student with 4 study hours, 6 sleep hours
pred = model.predict([[4,6]])
print("Pass exam?" , "Yes" if pred[0]==1 else "No")
```
