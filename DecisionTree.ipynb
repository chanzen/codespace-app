```python
from sklearn.tree import DecisionTreeClassifier
import numpy as np

# Example data: income, credit score → approve/reject
X = np.array([[60,700],[40,750],[40,650],[80,800]])
y = np.array([1,1,0,1])  # 1=approve, 0=reject

model = DecisionTreeClassifier()
model.fit(X, y)

# Predict for applicant with income=50, credit=720
pred = model.predict([[50,720]])
print("Loan approved?" , "Yes" if pred[0]==1 else "No")
```
