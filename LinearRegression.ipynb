```python
import numpy as np
from sklearn.linear_model import LinearRegression
X = np.array([[1000], [1200], [1500], [2000]]) # sqft [1 column 4 rows]
y = np.array([2000, 2500, 3000, 4000]) # price

model = LinearRegression()
model.fit(X,y) # Fit the model

print("Intercept: ", model.intercept_)
print("Slop: ", model.coef_)
#Formula: y=\beta _0+\beta _1x

pred = model.predict([[1800]])
print("Predicted price: ", pred[0])
```
